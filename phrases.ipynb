```python
from gensim.models.phrases import Phraser, Phrases
import pandas as pd
```

```python
tisch = pd.read_pickle("pickles/tisch.pickle")
sept = pd.read_pickle("pickles/sept.pickle")
strongs = pd.read_pickle('pickles/strongs.pickle')
bible = pd.concat([sept, tisch])
bible
```

```python
books = [book["str"] for (id_, book) in bible.groupby("book")]
books
```

```python
bigrams = Phrases(books, min_count=10, threshold=0, scoring="npmi")
bigram_phraser = Phraser(bigrams)
bigram_texts = bigram_phraser[books]
unique_bigrams = pd.Series([word for book in bigram_texts for word in book if "_" in word]).unique()
```

```python
pairs = pd.Series(unique_bigrams).str.split('_')
pd.concat([strongs.loc[pairs.str[0], 'def'], strongs.loc[pairs.str[1], 'def']], axis=1)
```
