```python
import pandas as pd
import numpy as np
```

```python
tisch = pd.read_pickle('tisch')
sept = pd.read_pickle('sept.pickle')
```

```python
tisch
```

```python
nameless = pd.concat(
    [tisch[tisch['str'].isna()],
    sept[sept['str'].isna()]]
)
```

```python
table = str.maketrans("","", ' .·,:;!()«»-·')
nameless['filt_text'] = nameless['text'].str.translate(table)
```

```python
nameless['filt_text'].value_counts()
```

```python
indexes = (~nameless['filt_text'].str.fullmatch(r'\w+'))
nameless[indexes]
```

```python
no_str_words = pd.Series(nameless['filt_text'].unique())
new_strongs = pd.Series(
    'C' + np.arange(len(no_str_words)).astype(str),
    index = no_str_words
)
```
