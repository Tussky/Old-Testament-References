```python
import pandas as pd

strongs = pd.read_pickle('pickles/strongs.pickle')
sept = pd.read_pickle("pickles/sept.pickle")
tisch = pd.read_pickle("pickles/tisch.pickle")
bible = pd.concat([sept, tisch], axis=0)
bible
```

```python
# nouns, verbs, and adjectives
interesting_words = bible[
    bible['rmac'].str.startswith('n-')
    | bible['rmac'].str.startswith('v-')
    | bible['rmac'].str.startswith('a-')
    | (bible['rmac'] == 'adv')]

verses = [verse['str'] for id_, verse in interesting_words.groupby(['book', 'chapter', 'verse'])]
```

```python
from gensim.models.doc2vec import TaggedDocument, Doc2Vec

tagged = [TaggedDocument(verse, [i]) for i, verse in enumerate(verses)]
```

```python
model = Doc2Vec(tagged, vector_size=300, window=10, min_count=1, workers=16)
```

```python
verses = [verse for id_, verse in interesting_words.groupby(['book', 'chapter', 'verse'])]
```

```python
for i, similarty in model.dv.most_similar(0):
    print(list(verses[i].iloc[0][['book', 'chapter', 'verse']]))
```
